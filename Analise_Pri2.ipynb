```python
import pandas as pd

# Load the CSV file to examine its contents
file_path = '/content/PEDE_PASSOS_DATASET_FIAP.csv'
data = pd.read_csv(file_path)

# Display the first few rows of the dataframe to understand its structure
data.head()
```

```python
# Load the CSV file again with the correct delimiter
data = pd.read_csv(file_path, delimiter=';')

# Display the first few rows to verify the data is loaded correctly
data.head()

```

```python
import matplotlib.pyplot as plt

# Plotting the distribution of grades for 2022
plt.figure(figsize=(12, 6))

# Português
plt.subplot(1, 3, 1)
plt.hist(data['NOTA_PORT_2022'].dropna(), bins=10, edgecolor='black')
plt.title('Distribuição de Notas de Português 2022')
plt.xlabel('Nota')
plt.ylabel('Frequência')

# Matemática
plt.subplot(1, 3, 2)
plt.hist(data['NOTA_MAT_2022'].dropna(), bins=10, edgecolor='black')
plt.title('Distribuição de Notas de Matemática 2022')
plt.xlabel('Nota')
plt.ylabel('Frequência')

# Inglês
plt.subplot(1, 3, 3)
plt.hist(data['NOTA_ING_2022'].dropna(), bins=10, edgecolor='black')
plt.title('Distribuição de Notas de Inglês 2022')
plt.xlabel('Nota')
plt.ylabel('Frequência')

plt.tight_layout()
plt.show()
```

```python
# Calculating the average of performance indicators for each year
indicators_2020 = data[['INDE_2020', 'IAA_2020', 'IEG_2020', 'IPS_2020', 'IDA_2020']].mean()
indicators_2021 = data[['INDE_2021', 'IAA_2021', 'IEG_2021', 'IPS_2021', 'IDA_2021']].mean()
indicators_2022 = data[['INDE_2022', 'IAA_2022', 'IEG_2022', 'IPS_2022', 'IDA_2022']].mean()

# Combine the averages into a DataFrame for plotting
performance_data = pd.DataFrame({
    '2020': indicators_2020,
    '2021': indicators_2021,
    '2022': indicators_2022
}).T

# Plotting the average performance indicators over the years
performance_data.plot(kind='bar', figsize=(10, 6))
plt.title('Indicadores de Desempenho Médio (2020-2022)')
plt.xlabel('Ano')
plt.ylabel('Média dos Indicadores')
plt.xticks(rotation=0)
plt.legend(title='Indicadores')
plt.show()
```

```python
# Convert columns to numeric, coercing errors to NaN to handle non-numeric values
for column in ['INDE_2020', 'IAA_2020', 'IEG_2020', 'IPS_2020', 'IDA_2020',
               'INDE_2021', 'IAA_2021', 'IEG_2021', 'IPS_2021', 'IDA_2021',
               'INDE_2022', 'IAA_2022', 'IEG_2022', 'IPS_2022', 'IDA_2022']:
    data[column] = pd.to_numeric(data[column], errors='coerce')

# Recalculate the average of performance indicators for each year
indicators_2020 = data[['INDE_2020', 'IAA_2020', 'IEG_2020', 'IPS_2020', 'IDA_2020']].mean()
indicators_2021 = data[['INDE_2021', 'IAA_2021', 'IEG_2021', 'IPS_2021', 'IDA_2021']].mean()
indicators_2022 = data[['INDE_2022', 'IAA_2022', 'IEG_2022', 'IPS_2022', 'IDA_2022']].mean()

# Combine the averages into a DataFrame for plotting
performance_data = pd.DataFrame({
    '2020': indicators_2020,
    '2021': indicators_2021,
    '2022': indicators_2022
}).T

# Plotting the average performance indicators over the years
performance_data.plot(kind='bar', figsize=(10, 6))
plt.title('Indicadores de Desempenho Médio (2020-2022)')
plt.xlabel('Ano')
plt.ylabel('Média dos Indicadores')
plt.xticks(rotation=0)
plt.legend(title='Indicadores')
plt.show()
```

Os indicadores de desempenho médio ao longo dos anos (2020-2022) mostram:

INDE (Indicador de Desempenho Escolar): Mantém-se estável ao longo dos anos, com uma ligeira tendência de aumento em 2022.
IAA (Indicador de Aproveitamento Acadêmico): Aumenta gradualmente, indicando uma melhora no aproveitamento acadêmico dos alunos.
IEG (Indicador de Eficiência de Gestão): Permanece relativamente constante, sugerindo uma consistência na eficiência da gestão educacional.
IPS (Indicador de Progresso do Sistema): Flutua levemente, mas não apresenta mudanças significativas.
IDA (Indicador de Desempenho Acadêmico): Mostra uma tendência de crescimento ao longo dos anos, destacando um avanço no desempenho acadêmico geral.

```python
# Filtrar dados para escolas públicas e privadas
public_schools = data[data['INSTITUICAO_ENSINO_ALUNO_2020'] == 'Escola Pública']
private_schools = data[data['INSTITUICAO_ENSINO_ALUNO_2020'] != 'Escola Pública']

# Calcular a média dos indicadores de desempenho para cada grupo de escolas
public_indicators_2022 = public_schools[['INDE_2022', 'IAA_2022', 'IEG_2022', 'IPS_2022', 'IDA_2022']].mean()
private_indicators_2022 = private_schools[['INDE_2022', 'IAA_2022', 'IEG_2022', 'IPS_2022', 'IDA_2022']].mean()

# Combinar as médias em um DataFrame para plotagem
school_comparison = pd.DataFrame({
    'Escolas Públicas': public_indicators_2022,
    'Escolas Privadas': private_indicators_2022
})

# Plotar comparação entre escolas públicas e privadas
school_comparison.plot(kind='bar', figsize=(10, 6))
plt.title('Comparação de Indicadores de Desempenho (2022)')
plt.xlabel('Indicadores')
plt.ylabel('Média dos Indicadores')
plt.xticks(rotation=0)
plt.legend(title='Tipo de Escola')
plt.show()
```

A comparação dos indicadores de desempenho entre escolas públicas e privadas em 2022 revela:

INDE (Indicador de Desempenho Escolar): As escolas privadas apresentam um desempenho ligeiramente superior ao das escolas públicas.
IAA (Indicador de Aproveitamento Acadêmico): As escolas privadas têm um aproveitamento acadêmico médio um pouco maior que o das escolas públicas.
IEG (Indicador de Eficiência de Gestão): O indicador é mais alto nas escolas privadas, indicando maior eficiência de gestão em comparação com as públicas.
IPS (Indicador de Progresso do Sistema): Similar para ambos os tipos de escolas, com uma ligeira vantagem para as escolas privadas.
IDA (Indicador de Desempenho Acadêmico): O desempenho acadêmico é melhor nas escolas privadas, mas a diferença não é muito grande.
As escolas privadas tendem a ter um desempenho ligeiramente superior em todos os indicadores analisados, o que pode refletir diferentes recursos, métodos de ensino, ou outras variáveis associadas ao tipo de instituição.

```python
# Calculate the yearly statistics for the IEG indicator
ieg_2020 = data['IEG_2020'].dropna()
ieg_2021 = data['IEG_2021'].dropna()
ieg_2022 = data['IEG_2022'].dropna()

# Calculate mean and standard deviation for each year
ieg_stats = pd.DataFrame({
    'Year': ['2020', '2021', '2022'],
    'Mean': [ieg_2020.mean(), ieg_2021.mean(), ieg_2022.mean()],
    'Standard Deviation': [ieg_2020.std(), ieg_2021.std(), ieg_2022.std()]
})

# Plotting the IEG variations over the years
plt.figure(figsize=(10, 6))
plt.errorbar(ieg_stats['Year'], ieg_stats['Mean'], yerr=ieg_stats['Standard Deviation'], fmt='-o', capsize=5)
plt.title('Variações do IEG ao Longo dos Anos')
plt.xlabel('Ano')
plt.ylabel('Média do IEG com Desvio Padrão')
plt.grid(True)
plt.show()

ieg_stats
```

A análise das variações do indicador IEG (Indicador de Eficiência de Gestão) ao longo dos anos mostra:

2020:

Média: 7,68
Desvio Padrão: 2,38
O desempenho foi relativamente alto, com uma variação moderada entre os alunos.
2021:

Média: 6,82
Desvio Padrão: 2,46
Houve uma queda na média, mas a variação entre os alunos permaneceu semelhante ao ano anterior.
2022:

Média: 7,88
Desvio Padrão: 1,65
A média do IEG aumentou, indicando uma melhora na eficiência de gestão, com uma redução na variação entre os alunos, sugerindo mais consistência.
Essas mudanças indicam que, após um declínio em 2021, houve uma recuperação significativa em 2022, acompanhada por uma redução na dispersão dos dados, o que pode refletir em uma aplicação mais uniforme das práticas de gestão eficiente.
